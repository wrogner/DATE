### Python is different

##### Python uses indenting (whitespace) to signal program structure (called scope)

```python
# The pound sign marks the start of a comment. Python itself
# ignores the comments, but they're helpful for anyone reading the code.
for i in [1, 2, 3, 4, 5]:
    """Strings starting with 3 \"\"\" or ''' can span
       multiple lines. They can be user for multiline string assignments
       and as multiline comments.
    Indentation is ignored by the Python interpreter."""

    print(f"i: {i}")                                    # first line in "for i" block
    for j in [1, 2, 3, 4, 5]:
        print(f"   j: {j}")                             # first line in "for j" block
        print(f" i+j: {i + j}")                         # last line in "for j" block
    print(f"_i: {i}")                                   # last line in "for i" block

print("done looping")
```

### Functions

Functions are encapsulated program instructions that can be called repeatedly. They take input parameters and provide output parameters.

```python
# a simple function that doubles its input

def double(x):
    """
    This is where you put an optional docstring that explains what the
    function does. For example, this function multiplies its input by 2.
    """
    return x * 2

double(2)
```

```python
# functions can be parameters

# apply_to_one takes a function as a parameter and calls it
def apply_to_one(f):
    """Calls the function f with 1 as its argument"""
    return f(1)                                                 # call f with parameter 1 (assuming that f takes one argument)

my_double = double                                              # refers to the previously defined function
x = apply_to_one(my_double)                                     # equals 2

assert x == 2
x
```

```python
# this can be extended

# apply_to_one takes a function as a parameter and calls it
def apply_to_any(f, param=1):

    assert type(param) == int, "param needs to be of type int"

    """Calls the function f with 1 as its argument"""
    return f(param)                                             # call f with param, default is 1


my_double = double                                              # refers to the previously defined function
x = apply_to_any(my_double, 2)                                  # equals 4

assert x == 4
x
```

```python
# Python offers simple functional programming
# This is carried out via anonymous functions, called lambda expressions

y = apply_to_one(lambda x: x + 4)                               # equals 5

assert y == 5
y
```

### Error handling with Exceptions

> &nbsp;  
> _All input is evil_
> <br><br>

Error handling can be done like in the old days by checking all parameter values and types.  
&rarr; This is fast but error prone.

Errors can be caught with initial `assert`s.  
&rarr; This catches errors but terminates the program in case it runs into one.

Errors can be handled using Exceptions.  
&rarr; This introduces a runtime penalty but increases program stability.

__If in doubt__ (and you should alway doubt your inputs), __wrap code in exceptions.__

```python
# this code will break

div_by_zero = 15 / 0
```

```python
# you can handle such conditions wrapping your code into a try - except phrase

try:
    div_by_zero = 15 / 0                                # this will fail for obvious reasons
except ZeroDivisionError:
    print("don't divide by 0")
except:                                                 # this is the default or catch-all phrase. Must be last
    print("Some error occured, but the programmer did not care about proper error handling")
```

```python
# lists can be 'sliced', a way to manipulate (extract, replace) list elements

first_three = x[:3]                                     # [-1, 1, 2]
three_to_end = x[3:]                                    # [3, 4, ..., 9]
one_to_four = x[1:5]                                    # [1, 2, 3, 4]
last_three = x[-3:]                                     # [7, 8, 9]
without_first_and_last = x[1:-1]                        # [1, 2, ..., 8]
copy_of_x = x[:]                                        # [-1, 1, 2, ..., 9]

every_third = x[::3]                                    # [-1, 3, 6, 9]
five_to_three = x[5:2:-1]                               # [5, 4, 3]

assert every_third == [-1, 3, 6, 9]
assert five_to_three == [5, 4, 3]

print(every_third)
print(five_to_three)
```

##### Tuples

Tuples are special in that they are not mutable once assigned.

```python
# difference between lists and tuples

my_list = [1, 2]
print(my_list)

my_tuple = (1, 2)
print(my_tuple)

other_tuple = 3, 4
print(other_tuple)

my_list[1] = 3                                          # my_list is now [1, 3]
print(my_list)

# retrieving indexed values from tuples works
print(my_tuple[1])

try:
    my_tuple[1] = 3
except TypeError as e:
    # cannot modify tuple
    print(f"{e.__class__.__name__}: {e}")
```

```python
# tuples are used to return multiple values

def sum_and_product(x, y):
    return (x + y), (x * y)

sp = sum_and_product(2, 3)                              # sp is (5, 6)
s, p = sum_and_product(5, 10)                           # s is 15, p is 50

assert sp == (5, 6)
assert s, p == (15, 50)

x, y = 1, 2                                             # now x is 1, y is 2
x, y = y, x                                             # Pythonic way to swap variables; now x is 2, y is 1

assert x == 2
assert y == 1
x, y
```

##### Dictionaries

Dictionaries are a collection of `key` - `value` pairs.

```python
empty_dict = {}                                         # Pythonic
empty_dict2 = dict()                                    # less Pythonic
grades = {"Joel": 80, "Tim": 95}                        # dictionary literal

joels_grade = grades["Joel"]                            # equals 80

assert joels_grade == 80
```

```python
# dicts cannot retrieve keys they don't contain

grades["Kate"]
```

```python
# checking for keys in a dictionary

print("Joel" in grades)

# keys are type sensitive
print("joel" in grades)
```

```python
# dictionaries can return default values using get

joels_grade = grades.get("Joel", 0)                     # equals 80
kates_grade = grades.get("Kate", 0)                     # equals 0
no_ones_grade = grades.get("No One")                    # default default is None

print(joels_grade)
print(kates_grade)
print(no_ones_grade)
```

```python
# dictionaries can be extended using assignments

grades["Tim"] = 99                                      # replaces the old value
grades["Kate"] = 100                                    # adds a third entry

print(grades)
print(len(grades))
```

```python
# create simple data structures with dictionaries

tweet = {
    "user": "joelgrus",
    "text": "Data Science is Awesome",
    "retweet_count": 100,
    "hashtags": ["#data", "#science", "#datascience", "#awesome", "#yolo"]
}

tweet_keys = tweet.keys()                               # iterable for the keys
print(tweet_keys)
tweet_values = tweet.values()                           # iterable for the values
print(tweet_values)
tweet_items = tweet.items()                             # iterable for the (key, value) tuples
print(tweet_items)

assert "user" in tweet_keys
assert "user" in tweet
assert "joelgrus" in tweet_values

print("user" in tweet_keys)                             # True, but not Pythonic
print("user" in tweet)                                  # Pythonic way of checking for keys
print("joelgrus" in tweet_values)                       # True (slow but the only way to check)
```

### Comprehensions

Comprehensions are a quick way in Python to iterate over arrays (container types) and select a subset or transform the array.

It can be used for generating `lists`, `dicts` and `tuples`.

```python
# list comprehensions

even_numbers = [x for x in range(5) if x % 2 == 0]      # [0, 2, 4]
squares = [x * x for x in range(5)]                     # [0, 1, 4, 9, 16]
even_squares = [x * x for x in even_numbers]            # [0, 4, 16]


assert even_numbers == [0, 2, 4]
assert squares == [0, 1, 4, 9, 16]
assert even_squares == [0, 4, 16]

print(even_numbers)
print(squares)
print(even_squares)
```

```python
# dict comprehensions

square_dict = {x: x * x for x in range(5)}              # {0: 0, 1: 1, 2: 4, 3: 9, 4: 16}
square_set = {x * x for x in [1, -1]}                   # {1} works with sets as well

assert square_dict == {0: 0, 1: 1, 2: 4, 3: 9, 4: 16}
assert square_set == {1}

print(square_dict)
print(square_set)
```

```python
zeros = [0 for _ in even_numbers]                       # has the same length as even_numbers

assert zeros == [0, 0, 0]
zeros
```

```python
# generate list of tuples
pairs = [(x, y)
         for x in range(10)
         for y in range(10)]   # 100 pairs (0,0) (0,1) ... (9,8), (9,9)

assert len(pairs) == 100
pairs
```

```python
# generate list of tuples with running index

increasing_pairs = [(x, y)                              # only pairs with x < y,
                    for x in range(10)                  # range(lo, hi) equals
                    for y in range(x + 1, 10)]          # [lo, lo + 1, ..., hi - 1]


assert len(increasing_pairs) == 9 + 8 + 7 + 6 + 5 + 4 + 3 + 2 + 1
assert all(x < y for x, y in increasing_pairs)          # use comprehension to generate test tuples
increasing_pairs
```

### Classes

Classes provide object oriented programming in Python.

```python
# define a class that counts people in a plane or concert
# each time a person enters, click
# the counter will keep track of the number of people

class CountingClicker:
    """A class can/should have a docstring, just like a function"""

    # class constructor
    def __init__(self, count=0):
        self.count = count

    # standard representation if printed
    def __repr__(self):
        return f"CountingClicker(count={self.count})"

    # class methods
    def click(self, num_times=1):
        """Click the clicker some number of times."""
        self.count += num_times

    def read(self):
        return self.count

    def reset(self):
        self.count = 0


clicker = CountingClicker()
assert clicker.read() == 0, "clicker should start with count 0"
clicker.click()
clicker.click()

assert clicker.read() == 2, "after two clicks, clicker should have count 2"

print(clicker)                                          # this calls __repr__

clicker.reset()
assert clicker.read() == 0, "after reset, clicker should be back to 0"
```

```python
clicker = CountingClicker(100)                          # preset with 100
clicker = CountingClicker(count=100)                    # init with keyword

clicker.click()
clicker.read()
```

```python
# A subclass inherits all the behavior of its parent class.

class NoResetClicker(CountingClicker):
    # This class has all the same methods as CountingClicker

    # Except that it has a reset method that does nothing.
    def reset(self):                                    # override reset
        pass


clicker2 = NoResetClicker()
assert clicker2.read() == 0
clicker2.click()
assert clicker2.read() == 1
clicker2.reset()
assert clicker2.read() == 1, "reset shouldn't do anything"
clicker2
```

### Generators

Storing large arrays in lists or dicts requires large amounts of ram. In some cases, providing the data can be deferred to the point where data is accessed.

This is provided by generators in Python.

```python
# define a generator

def generate_range(n):
    i = 0
    while i < n:
        yield i                                             # every call to yield produces a value of the generator
        i += 1

for i in generate_range(10):                                # each call to generate_range will return an integer
    print(f"i: {i}")                                        # that is generated on the fly

generate_range(5)
```

```python
# generators can be stacked. In this case, a comprehension wrappes around a generator

evens_below_20 = (i for i in generate_range(20) if i % 2 == 0)
evens_below_20
```

```python
def natural_numbers():
    """returns 1, 2, 3, ..."""
    n = 1
    while True:
        yield n
        n += 1
```

```python
# None of these computations *does* anything until we iterate

data = natural_numbers()
evens = (x for x in data if x % 2 == 0)
even_squares = (x ** 2 for x in evens)
even_squares_ending_in_six = (x for x in even_squares if x % 10 == 6)
# and so on

# however, calling them with next, generates a result

assert next(even_squares_ending_in_six) == 16
assert next(even_squares_ending_in_six) == 36
assert next(even_squares_ending_in_six) == 196

next(data)
```

##### Getting value and index at the same time

```python
# if we need a reference counter

names = ["Alice", "Bob", "Charlie", "Debbie"]

# not Pythonic
for i in range(len(names)):
    print(f"name {i} is {names[i]}")

# also not Pythonic
i = 0
for name in names:
    print(f"name {i} is {names[i]}")
    i += 1

# Pythonic
for i, name in enumerate(names):
    print(f"name {i} is {name}")
```

```python
%%timeit
for i in range(len(names)):
    print(f"name {i} is {names[i]}")
```

```python
%%timeit
i = 0
for name in names:
    print(f"name {i} is {names[i]}")
    i += 1
```

```python
%%timeit
for i, name in enumerate(names):
    print(f"name {i} is {name}")
```

The pythonic way usually is the fastest!

### zip and argument unpaching

```python
# zip is used to pairwise combine two arrays until one exhausts

list1 = ['a', 'b', 'c', 'd']                            # mind list1 is longer
list2 = [1, 2, 3]

# zip is lazy (a generator), so you have to do something like the following
pairs = [pair for pair in zip(list1, list2)]            # is [('a', 1), ('b', 2), ('c', 3)]

assert pairs == [('a', 1), ('b', 2), ('c', 3)]
pairs
```

```python
# argument unpacking happens using *-notation

pairs = [('a', 1), ('b', 2), ('c', 3)]
letters, numbers = zip(('a', 1), ('b', 2), ('c', 3))
print(letters, numbers)

# unpack using *pairs
letters, numbers = zip(*pairs)

print(letters, numbers)
```

```python
%%timeit
letters, numbers = zip(('a', 1), ('b', 2), ('c', 3))
```

```python
%%timeit
letters, numbers = zip(*pairs)
```
