- lectures on key topics (25%)

- practical exercises with Python or Jupyter (25%)

### Learning results

After successfully completing this course, students ...

- can utilize Python, Jupyter and selected scientific libraries for data analysis
- can collect, prepare, statistically analyze and visualize relevant data
- can apply fundamental methods and procedures of data analysis to selected problems
- can research, develop and apply advanced data analysis topics based on individual demand

- Introduction to Python, Jupyter, selected scientific libraries and visualization
